## Lists

### A Python list is similar to an array in other languages. In Python, an empty list can be created in the following ways. 


```python
list=[]
```

```python
my_list = [1, 2, 3]
my_list2 = ["a", "b", "c"]
my_list3 = ["a", 1, "Python", 5]
len(my_list)
```

### The first list has 3 integers, the second has 3 strings and the third has a mixture. You can also create lists of lists like this:



```python
my_nested_list = [my_list, my_list2]
print(my_nested_list)
print("The length of nested list is: ",len(my_nested_list))
```

### Occasionally, you’ll want to combine two lists together,  just add two lists together.

```python
my_list = [1, 2, 3]
my_list2 = ["a", "b", "c"]
combo_list = my_list + my_list2
combo_list
```

### You can also sort a list. Let’s spend a moment to see how to do that:



```python
alpha_list = [34, 23, 67, 100, 88, 2]
alpha_list.sort()
print(alpha_list)
print(len(alpha_list))
```

### You can slice a list 

```python
alpha_list[:-2]
```

### append(): Adds an element to the end of the list.

```python
my_list = [1, 2, 3]
my_list.append(5)
print(my_list)
```

```python
a=[]
for i in range(10):
    a.append(i)
a
```

### extend(): Extends the list by appending elements from the iterable.

```python
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)  # Output: [1, 2, 3, 4]
```

### insert(): Inserts an element at the specified position.

```python
my_list = [1, 2, 3]
my_list.insert(1, 5)
print(my_list)  # Output: [1, 5, 2, 3]
```

### remove(): Removes the first occurrence of a value from the list.

```python
my_list = [1, 2, 3, 4, 3]
my_list.remove(3)
print(my_list) 
```

# Tuples

## A tuple is similar to a list, but you create them with parentheses instead of square brackets.

## The main difference is that a tuple is immutable while the list is mutable

### Immutability: Tuples are immutable, ensuring data integrity by preventing accidental changes.

#### Performance: Tuples are often more memory-efficient and faster to iterate over than lists, making them suitable for large datasets.

#### Hashability: Tuples can be used as keys in dictionaries and elements in sets, facilitating efficient data organization and lookup.

#### Data Integrity: Immutable nature of tuples maintains data consistency, crucial in AI applications.

#### Semantic Clarity: Tuples provide clear grouping of related elements, enhancing code readability and maintainability.

## Dictionaries

### dictionary is an unordered set of key:value pairs and the keys must be unique. You can get a list of keys by calling a dictionary instance’s keys method. To check if a dictionary has a key, you can use Python’s in keyword. 


### The first two examples show how to create an empty dictionary.

```python
my_dict = {}
# another_dict = dict()
my_other_dict = {"one":6, "two":2, "three":3}
my_other_dict={'three': 3, 'two': 2, 'one': 5}
my_other_dict
```

### Lets check how to access a value in a dictionary.

```python
print(my_other_dict["one"])
my_dict = {"name":"Mike", "address":"123 Happy Way"}
my_dict["name"]
```

### If you need to get a listing of all the keys in a dictionary, then you do this:

```python
my_dict.keys()
```

### Print the length of the dictionary

```python
len(my_dict)
```

### Example that demonstrates the use of lists, tuples, and dictionaries in an AI context:

```python
# List: Storing Dataset Features
features = ['age', 'gender', 'income']

# Tuple: Model Hyperparameters
hyperparameters = (0.01, 100, 32) 

# Dictionary: Feature-Value Pairs for a Data Point
data_point = {'age': 30, 'gender': 'male', 'income': 50000}

# Accessing elements
print("First feature:", features[0])  # Output: 'age'
print("Number of epochs:", hyperparameters[1])  # Output: 100
print("Income of the data point:", data_point['income'])  # Output: 50000

# Modifying elements
features.append('education')
print("Updated features:", features)  # Output: ['age', 'gender', 'income', 'education']

# Attempting to modify a tuple will result in an error
# hyperparameters[0] = 0.001  # Raises TypeError: 'tuple' object does not support item assignment

# Modifying values in the dictionary
data_point['income'] = 60000
print("Updated data point:", data_point)  # Output: {'age': 30, 'gender': 'male', 'income': 60000}
```
