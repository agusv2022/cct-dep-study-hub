```python
# Import the libraries
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("arbuthnot.txt", sep = r"\s+")

df.head()
```

```python
# Check the missing values
df.isnull().sum()
```

```python
# Check the duplicate values
print(df.duplicated().sum())
```

```python
# Minimum and maximum values of the year
print("First year: ", df["year"].min())
print("Last year: ", df["year"].max())
```

```python
# Total number of rows and columns
df.shape
```

```python
# Detailed information of the dataset
df.info()
```

```python
# Basic statistics
df.describe()
```

```python
import matplotlib.pyplot as plt

# Calculate boys-to-girls ratio
df["ratio"] = df["boys"]/df["girls"]

df[["year", "ratio"]]
```

```python
# Size of the Figure
plt.figure(figsize = (8, 4))

# plot function
plt.plot( df["year"], df["ratio"], marker = "o", markersize = 3 )

# Reference when the boys and girls birth are equal
plt.axhline(y = 1, linestyle = "--")

plt.title("Boys-to-Girls Ratio in the London from 1629-1710")
plt.xlabel("Year")
plt.ylabel("Boys-to-Girls Ratio")
plt.grid(True)
```
