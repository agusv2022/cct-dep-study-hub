```python
# x = 10
# y = 25
x = int(input())
y = int(input())
sum = x + y
sub = x - y
mult = x * y
div = x / y
remainder = x % y
power = x ** y
print('The sum of {} and {} is equal to: {}'.format(x, y, sum))
print('The subtraction of {} and {} is equal to: {}'.format(x, y, sub))
print('The multiplication of {} and {} is equal to: {}'.format(x, y, mult))
print('The div of {} and {} is equal to: {}'.format(x, y, div))
print('The remainder of {} and {} is equal to: {}'.format(x, y, remainder))
print('The power of {} and {} is equal to: {}'.format(x, y, power))
```
